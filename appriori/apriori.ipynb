```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
 
```

```python
data = pd.read_csv('sample_transactions.csv')
transactions = data.groupby('TransactionID')['Item'].apply(list).tolist()
 
```

```python
encoder = TransactionEncoder()
encoded_array = encoder.fit(transactions).transform(transactions)
df = pd.DataFrame(encoded_array, columns=encoder.columns_)
```

```python

# Find frequent itemsets 
frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)
print("Frequent Itemsets:\n", frequent_itemsets, "\n")
```

```python
# association rules 
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)
rules = rules.sort_values(by="lift", ascending=False)
```

```python
print("Association Rules:\n")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
